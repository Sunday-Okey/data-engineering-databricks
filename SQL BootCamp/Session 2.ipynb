```sql
-- Fetch male employees who earn more than 5000 salary.
SELECT *
FROM Employees
WHERE gender = 'M' AND Salary > 5000
```

```sql
-- Fetch employees whose name starts with J or whose salary is greater than or equal to 70000.
SELECT *
FROM Employees
WHERE Emp_Name LIKE 'J%' OR Salary >= 70000;
```
