```python
%sql
    USE CATALOG dbacademy;
    USE SCHEMA labuser14586003_1778537625;
```

```python
from pyspark.sql.functions import *
```

```python
df = spark.readStream.table('sourcetable')
```

```python
def myfunc(df,batch_id):

    df = df.groupBy("color").agg(count("*").alias("count"))

    # destination 1
    df.write.format("delta")\
            .mode("append")\
            .option("path","/Volumes/dbacademy/labuser14586003_1778537625/stream/dest1")\
            .save()

    # Destination 2 
    df.write.format("delta")\
            .mode("append")\
            .option("path","/Volumes/dbacademy/labuser14586003_1778537625/stream/dest2")\
            .save()
```

```python
df.writeStream.foreachBatch(myfunc)\
    .outputMode("append")\
    .trigger(once=True)\
    .option("checkpointLocation","/Volumes/dbacademy/labuser14586003_1778537625/stream/foreachsink/checkpoint")\
    .start()
```
