```python
%sql
    USE CATALOG dbacademy;
    USE SCHEMA labuser14586003_1778537625;
```

```python
%sql
CREATE TABLE windowtbl
(
  color STRING,
  event_date TIMESTAMP
)
```

```python
%sql
INSERT INTO windowtbl
VALUES
('red',"2026-05-12T02:15:00.000Z")
```

```python
df = spark.readStream.table("windowtbl")
```

```python
from pyspark.sql.functions import * 
```

```python
df = df.groupBy("color",window("event_date","10 minutes"))\
            .agg(count(lit(1)).alias("color_count"))
```

```python
df.writeStream.format("delta")\
        .outputMode("complete")\
        .trigger(once=True)\
        .option("path","/Volumes/dbacademy/labuser14586003_1778537625/stream/windows")\
        .option("checkpointLocation","/Volumes/dbacademy/labuser14586003_1778537625/stream/check_windows")\
        .start()
```
